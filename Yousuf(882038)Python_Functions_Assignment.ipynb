## Question 12 — Electricity Bill Function

**Problem:**  
An electricity company charges different rates based on units consumed. Create a function that calculates the bill.

**Suggested rates (you may adjust slightly):**
- 0 – 100 units → Rs. 5 per unit
- 101 – 200 units → Rs. 10 per unit
- Above 200 units → Rs. 15 per unit

**Requirements:**
- Create a function that accepts the number of **units consumed**.
- Calculate and **return** the final electricity bill using conditions.
- Display the returned bill outside the function.

**Focus:** Function + arguments + conditions + `return`

````python
def electricity_bill(units):
    if units <= 100:
        bill = units * 5
    elif units <= 200:
        bill = (100 * 5) + ((units - 100) * 10)
    else:
        bill = (100 * 5) + (100 * 10) + ((units - 200) * 15)
    return bill

bill = electricity_bill(250)
print("Electricity Bill: Rs.", bill)
````

**Problem:**

## Question 16 — Student Marks Processor

**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

**Requirements:**
- Create a function that accepts a **list of student marks**.
- Using a loop (or appropriate logic), calculate and **return**:
  - Highest mark
  - Lowest mark
  - Total marks
  - Average marks
- Display the results outside the function.

**Focus:** Function + list + loop + `return`

````python
def marks_processor(marks):
    if len(marks) == 0:
        return None, None, 0, 0

    total = 0
    highest = marks[0]
    lowest = marks[0]

    for mark in marks:
        total += mark
        if mark > highest:
            highest = mark
        if mark < lowest:
            lowest = mark

    average = total / len(marks)
    return highest, lowest, total, average

marks = [75, 88, 92, 65, 80]
highest, lowest, total, average = marks_processor(marks)

print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print("Total Marks:", total)
print("Average Marks:", average)
````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
def process_order(price, quantity, discount_percentage, tax_percentage):
    def calculate_subtotal():
        return price * quantity

    def calculate_discount(subtotal):
        return subtotal * discount_percentage / 100

    def calculate_tax(amount):
        return amount * tax_percentage / 100

    def calculate_final_amount():
        subtotal = calculate_subtotal()
        discount = calculate_discount(subtotal)
        amount_after_discount = subtotal - discount
        tax = calculate_tax(amount_after_discount)
        return amount_after_discount + tax

    return calculate_final_amount()

final_amount = process_order(2000, 3, 10, 5)
print("Final Order Amount: Rs.", final_amount)
````

  - Final bill

## Question 29 — Student Result Management System

**Problem:**  
Create a complete student result system using functions.

**The system should:**
1. Accept student information
2. Accept multiple subject marks
3. Calculate total
4. Calculate percentage
5. Determine grade
6. Determine pass/fail
7. Display a final result card

**Requirements:**
- Use **multiple functions**
- Use arguments and `return`
- Use lists
- Use conditions and loops
- Use at least one advanced concept (`*args`, `**kwargs`, nested functions, or lambda)
- Do **not** write the entire program inside one function

**Focus:** Combining multiple function concepts

````python
def student_information(name, roll_number):
    return {"Name": name, "Roll Number": roll_number}

def calculate_total(marks):
    total = 0
    for mark in marks:
        total += mark
    return total

def calculate_percentage(total, total_marks):
    return (total / total_marks) * 100

def calculate_grade(percentage):
    if percentage >= 80:
        return "A+"
    elif percentage >= 70:
        return "A"
    elif percentage >= 60:
        return "B"
    elif percentage >= 50:
        return "C"
    elif percentage >= 40:
        return "D"
    else:
        return "F"

def check_pass_fail(percentage):
    return "Pass" if percentage >= 40 else "Fail"

student = student_information("Ali Khan", "STU101")
subject_marks = [85, 78, 90, 75, 82]
total = calculate_total(subject_marks)
percentage = calculate_percentage(total, 500)
grade = calculate_grade(percentage)
status = check_pass_fail(percentage)

print("========== STUDENT RESULT CARD ==========")
for key, value in student.items():
    print(key + ":", value)
print("Subject Marks:", subject_marks)
print("Total Marks:", total, "/ 500")
print("Percentage:", round(percentage, 2), "%")
print("Grade:", grade)
print("Result:", status)
print("=========================================")
````

````python
def customer_info(name, address):
    return {"Customer Name": name, "Address": address}

def product_details():
    return [
        {"name": "Keyboard", "price": 2500, "quantity": 1},
        {"name": "Mouse", "price": 1000, "quantity": 2},
        {"name": "Headphones", "price": 3000, "quantity": 1}
    ]

def calculate_item_total(price, quantity):
    return price * quantity

def calculate_subtotal(*totals):
    return sum(totals)

def calculate_discount(subtotal, discount_percentage=10):
    return subtotal * discount_percentage / 100

def calculate_tax(amount, tax_percentage=5):
    return amount * tax_percentage / 100

def calculate_delivery(subtotal):
    if subtotal >= 5000:
        return 0
    return 250

def calculate_final_bill(subtotal, discount, tax, delivery):
    # Nested function to calculate the payable amount
    def get_final_amount():
        return subtotal - discount + tax + delivery
    return get_final_amount()

def show_order_summary(customer, products, subtotal, discount, tax, delivery, final_amount):
    print("\n========== E-COMMERCE ORDER SUMMARY ==========")
    for key, value in customer.items():
        print(key + ":", value)

    print("\nOrdered Products:")
    for product in products:
        print(product["name"], "-", product["quantity"], "x Rs.", product["price"],
              "= Rs.", product["total"])

    print("----------------------------------------------")
    print("Subtotal: Rs.", subtotal)
    print("Discount: Rs.", discount)
    print("Tax: Rs.", tax)
    print("Delivery Charges: Rs.", delivery)
    print("Final Amount: Rs.", final_amount)
    print("==============================================")

customer = customer_info("Ali Khan", "Karachi")
products = product_details()

item_totals = []
for product in products:
    product["total"] = calculate_item_total(product["price"], product["quantity"])
    item_totals.append(product["total"])

# Lambda function to calculate the total of all product amounts
get_total = lambda values: sum(values)
subtotal = get_total(item_totals)

discount = calculate_discount(subtotal)
amount_after_discount = subtotal - discount
tax = calculate_tax(amount_after_discount)
delivery = calculate_delivery(subtotal)
final_amount = calculate_final_bill(subtotal, discount, tax, delivery)

show_order_summary(customer, products, subtotal, discount, tax, delivery, final_amount)
````
